```python
import os
import random
```

```python
def thanos_project(path, num_files):
    if not os.path.exists(path):
        os.makedirs(path)

    for i in range(1, num_files + 1):
        file_name = "text_" + str(i) + ".txt"
        file_path = os.path.join(path, file_name)
        f = open(file_path, "a")
        f.close()
    
    file_names = os.listdir(path)
    print("Number of files before deletion:", len(file_names))

    half = len(file_names) // 2
    files_to_delete = random.sample(file_names, half)

    for file_name in files_to_delete:
        file_path = os.path.join(path, file_name)
        os.remove(file_path)

    file_names_after = os.listdir(path)
    print("Number of files after deletion:", len(file_names_after))


thanos_project("tmp_directory", 20)
```
