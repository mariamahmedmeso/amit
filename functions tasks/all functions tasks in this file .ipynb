```python
def multiply(num):
    for i in range(1, 11):
        print(f"{num} * {i} = {num * i}")
multiply(9)
```

```python
def prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
def twin_primes(limit):
    print("Twin Prime:")
    for n in range(2, limit - 2):
        if prime(n) and prime(n + 2):
            print(f"{n} and {n + 2}")
twin_primes(1000)
```

```python
def prime(num):
    factors = []
    d = 2
    while d * d <= num:
        while num % d == 0:
            factors.append(d)
            num //= d
        d += 1
    if num > 1:
        factors.append(num)
    return factors
print(prime(56))
```

```python
def dectobin(num):
    if num == 0:
        return "0"
    result = ""
    while num > 0:
        result = str(num % 2) + result
        num //= 2
    return result

print(dectobin(11))
```

```python
def perfect_nums(start, end):
    for num in range(start, end + 1):
        divisors_sum = sum(i for i in range(1, num) if num % i == 0)
        if divisors_sum == num and num != 0:
            print(num)

perfect_nums(0, 100)
```

```python
def get_number(prompt):
    while True:
        value = input(prompt)
        try:
            return float(value)
        except ValueError:
            print("Please enter a valid number.")
def calculator():
    while True:
        print("Select an operation")
        print("1.(+)")
        print("2.(-)")
        print("3.(*)")
        print("4.(/)")
        choice = input("Enter your choice (1/2/3/4) or 'exit' to quit: ")
        if choice == "exit":
            print("Exiting the calculator")
            break
        if choice not in ("1", "2", "3", "4"):
            print("Invalid choice, try again.")
            continue
        num1 = get_number("Enter first number: ")
        num2 = get_number("Enter second number: ")
        if choice == "1":
            result = num1 + num2
            print(f"adding {num1} and {num2} is {result}")
        elif choice == "2":
            result = num1 - num2
            print(f"subtracting {num2} from {num1} is {result}")
        elif choice == "3":
            result = num1 * num2
            print(f" multiplying {num1} and {num2} is {result}")
        elif choice == "4":
            if num2 == 0:
                print("Cannot divide by zero, try again.")
                continue
            result = num1 / num2
            print(f"{num1} / {num2} = {result}")

calculator()
```

```python
import os
import random
def thanos_project(folder_path, files_count):
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
    for i in range(files_count):
        file_path = os.path.join(folder_path, f"file_{i}.txt")
        with open(file_path, "w") as f:
            f.write("")
    files = os.listdir(folder_path)
    print(f"Number of files before deletion: {len(files)}")
    half = len(files) // 2
    to_delete = random.sample(files, half)
    for file_name in to_delete:
        os.remove(os.path.join(folder_path, file_name))
    remaining_files = os.listdir(folder_path)
    print(f"Number of files after deletion: {len(remaining_files)}")
```
