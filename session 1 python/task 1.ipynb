```python
email = input()
if email.count("@") != 1:
    print("invalid email")
else:
    a = email.index("@")
    b = email.rfind(".")
    if b < a:
        print("invalid email")
    else:
        print(email[:a])
        print(email[a + 1:b])
        if email.endswith(".com"):
            print("commercial domain")
        elif email.endswith(".edu"):
            print("educational domain")
        else:
            print("other domain")
```

```python
s = "###!!@mocleW EPGTQ!!!6789"
x = ""
for i in s:
    if i.isalpha() or i == " ":
        x += i
a = x.split()
w1 = a[0][::-1]
w2 = a[1]
print(w1, w2)
```
