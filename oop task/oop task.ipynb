prime twin

```python
class twin_primes:
    def __init__(self, limit):
        self.limit = limit

    def prime(self, n):
        
        if n < 2:
            return False

        for i in range(2, int(n ** 0.5) + 1):
            if n % i == 0:
                return False

        return True

    def find_twin_primes(self):
       
        print("twin prime:")

        for n in range(2, self.limit - 2):
            if self.prime(n) and self.prime(n + 2):
                print(f"{n} and {n + 2}")
```

```python
numbers = twin_primes(1000)

numbers.find_twin_primes()
```

prime factor

```python
class prime_factors:
    def __init__(self, num):
        self.num = num

    def find_factors(self):
        factors = []
        d = 2
        num = self.num

        while d * d <= num:
            while num % d == 0:
                factors.append(d)
                num //= d
            d += 1
        if num > 1:
            factors.append(num)
        return factors
```

```python
numbers = prime_factors(56)

print(numbers.find_factors())
```

```python
class decimal_to_binary:

    def __init__(self, num):
        self.num = num

    def convert(self):
        if self.num == 0:
            return "0"
        result = ""
        num = self.num
        while num > 0:
            result = str(num % 2) + result
            num //= 2
        return result
```

```python
number = decimal_to_binary(11)
print(number.convert())
```

perfect numbers

```python
class perfect_numbers:

    def __init__(self, start, end):
        self.start = start
        self.end = end
    def find_perfect_numbers(self):
        for num in range(self.start, self.end + 1):
            divisors_sum = sum(i for i in range(1, num) if num % i == 0)

            if divisors_sum == num and num != 0:
                print(num)
```

```python
numbers = perfect_numbers(0, 100)
numbers.find_perfect_numbers()
```

calculation

```python
class calculator:
  
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def subtract(self):
        return self.num1 - self.num2

    def multiply(self):
        return self.num1 * self.num2

    def divide(self):
        if self.num2 == 0:
            return "cannot divide by zero"
        return self.num1 / self.num2
```

```python
while True:
    print("1. addition")
    print("2. subtraction")
    print("3. multiplication")
    print("4. division")
    print("5. exit")

    choice = input("choose an operation: ")

    if choice == "5":
        print("thank you for using the calculator")
        break

    try:
        num1 = float(input("enter the first number: "))
        num2 = float(input("enter the second number: "))
    except ValueError:
        print("please enter valid numbers")
        continue

    calc = calculator(num1, num2)

    if choice == "1":
        print("the result of adding", num1, "and", num2, "is", calc.add())

    elif choice == "2":
        print("the result of subtracting", num2, "from", num1, "is", calc.subtract())

    elif choice == "3":
        print("the result of multiplying", num1, "and", num2, "is", calc.multiply())

    elif choice == "4":
        print("the result of dividing", num1, "by", num2, "is", calc.divide())

    else:
        print("invalid choice")
        continue

    again = input("do you want another calculation? ")

    if again.lower() == "no":
        print("thank you for using the calculator")
        break
```

thanos project

```python
import os
import random
```

```python
class thanos_project:
    def __init__(self, folder_path, files_count):
        self.folder_path = folder_path
        self.files_count = files_count

    def create_files(self):
        if not os.path.exists(self.folder_path):
            os.makedirs(self.folder_path)

        for i in range(self.files_count):
            file_path = os.path.join(self.folder_path, f"file_{i}.txt")

            with open(file_path, "w") as f:
                f.write("")

    def delete_files(self):
        files = os.listdir(self.folder_path)

        print("number of files before deletion:", len(files))

        half = len(files) // 2

        to_delete = random.sample(files, half)

        for file_name in to_delete:
            os.remove(os.path.join(self.folder_path, file_name))

        remaining_files = os.listdir(self.folder_path)

        print("number of files after deletion:", len(remaining_files))
```

```python
project = thanos_project("test_folder", 10)
project.create_files()
project.delete_files()
```
